```python
subtitle_path='/home/hg22723/vseek/dataset/videomme_window_8'
import os

directories = os.listdir(subtitle_path)
count = 0
total_count = 0
for directory in directories:
    if os.path.isdir(os.path.join(subtitle_path, directory)):
        files_list = os.listdir(os.path.join(subtitle_path, directory))
        for file in files_list:
            file_path = os.path.join(subtitle_path, directory, file)
            if os.path.isfile(file_path):
                if 'subtitle_embeddings.pt' in file_path:
                    count += 1
                total_count += 1
            elif os.path.isdir(file_path):
                files_list = os.listdir(file_path)
                for file_ in files_list:
                    if 'subtitle_embeddings.pt' in file_:
                        count += 1
                    total_count += 1
            

print(count)
print(total_count)


```
