```python
import os
import json
import cv2
from decord import VideoReader, cpu
json_path = "/nas/mars/dataset/longvideobench/LongVideoBench/lvb_val.json"
subtitle_path = "/nas/mars/dataset/longvideobench/LongVideoBench/subtitles/{}"

with open(json_path, "r") as f:
    data = json.load(f)

# Non zero starting subtitle timestamp
# Portion of questions where starting stimestamp is greter tan video
non_zero_start_sub_count = 0
incorrect_start_sub_count = 0
for item in data:
    if item["starting_timestamp_for_subtitles"] > 0:
        non_zero_start_sub_count += 1
    if item["starting_timestamp_for_subtitles"] > item["duration"]:
        incorrect_start_sub_count += 1
        with open(subtitle_path.format(item["subtitle_path"]), "r") as f:
            subtitles = json.load(f)
        for sub in subtitles:
            print(f"starting timestamp: {sub['start']}, ending timestamp: {sub['end']}, starting timestamp for subtitles: {item['starting_timestamp_for_subtitles']}, duration: {item['duration']}")
    
        # read the video and output the duration
        video_path = "/nas/mars/dataset/longvideobench/LongVideoBench/videos/{}".format(item["video_path"])
        vr = VideoReader(video_path, ctx=cpu(0))
        fps = float(vr.get_avg_fps()) if hasattr(vr, 'get_avg_fps') else float(vr.get_avg_fps())
        num_frames = len(vr)
        duration_sec = num_frames / fps if fps > 0 else 0.0
        print(f"Duration: {duration_sec:.3f}s")

print(non_zero_start_sub_count / len(data))
print(incorrect_start_sub_count / len(data))

```
